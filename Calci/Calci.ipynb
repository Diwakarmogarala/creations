## Building a Calculator

```python
# Function
#### Addition
def add(a,b):
    return a+b

def subtract(a,b):
    return a-b

def multiply(a,b):
    return a*b

def divide(a,b):
    if b == 0:
        return "This was not possible"
    else:
        return a/b
```

# Calculator

```python

def calci():
    z = "yes"
    while True:
        if z == "yes":
            print("Type : 1 for add,\n       2 for subtract,\n       3 for multiply and\n       4 for divide")
            y = int(input(" Select option : "))
            if y>=5:
                continue
            a = float(input("Enter first value : "))
            b = float(input("Enter Second value : "))
            
            if y == 1:
                print(' {} + {}'.format(a,b)," = ", add(a,b))
            elif y == 2:
                print(' {} - {}'.format(a,b)," = ", subtract(a,b))
            elif y == 3:
                print(' {} * {}'.format(a,b)," = ", multiply(a,b))
            elif y == 4:
                print(' {} / {}'.format(a,b)," = ", divide(a,b))
            print("If you want to continue type : yes\nIf you want to exit type : bye")
            z = input("Type your response : ")
        if z == "bye":
            break
        
        
```
